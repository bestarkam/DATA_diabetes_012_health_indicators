```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8')
```

```python
# dataset
df = pd.read_csv(r"D:\MACHINE_LERNING_PROJECTS\DATA_diabetes_012_health_indicators\db\diabetes_012_health_indicators_BRFSS2015.csv")
df.head()
df.info()
df.describe()
```

```python
df_size = df.size
df_size
```

```python

# Valeurs manquantes
df.isnull().sum()

# Suppression simple si nécessaire
df = df.dropna()

# Vérification doublons
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
a = len(df)
a
```

```python
# Univariate Analysis
```

```python
# Histogramme BMI

plt.figure()

ax = sns.histplot(df['BMI'], bins=30)

plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")

# Ajouter les valeurs sur chaque barre
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.text(
            p.get_x() + p.get_width() / 2,
            height,
            int(height),
            ha='center',
            va='bottom',
            fontsize=8
        )

plt.show()
```

```python
# Histogramme age
plt.figure(figsize=(8,5))

ax = sns.histplot(df['Age'], bins=20)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

# Ajouter les valeurs sur les barres
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.text(
            p.get_x() + p.get_width() / 2,
            height,
            int(height),
            ha='center',
            va='bottom',
            fontsize=8
        )

plt.show()
```

```python
# Class Distribution
plt.figure(figsize=(6,4))

ax = sns.countplot(x='Diabetes_012', data=df)

plt.title("Diabetes Distribution")
plt.xlabel("Diabetes Status")
plt.ylabel("Count")

# Ajouter les valeurs sur les barres
for p in ax.patches:
    height = p.get_height()
    ax.text(
        p.get_x() + p.get_width() / 2,
        height,
        int(height),
        ha='center',
        va='bottom'
    )

plt.show()

(df['Diabetes_012'].value_counts(normalize=True) * 100).round(2)
```

```python
print("len(df):", len(df))
print("sum countplot:", sum([p.get_height() for p in ax.patches]))
print("NaN:", df['Diabetes_012'].isnull().sum())
print(df['Diabetes_012'].value_counts(dropna=False))
```

```python
# Bivariate Analysis
```

```python
# BMI vs Diabetes
fig, axes = plt.subplots(1, 2, figsize=(12,5))

# Graph
sns.boxplot(x='Diabetes_012', y='BMI', data=df, ax=axes[1])
axes[1].set_title("BMI vs Diabetes")

# Table
summary = df.groupby('Diabetes_012')['BMI'].agg(['mean', 'median', 'std'])
axes[0].axis('off')
axes[0].table(
    cellText=summary.round(2).values,
    colLabels=summary.columns,
    rowLabels=summary.index,
    loc='center'
)

axes[0].set_title("BMI Summary Statistics")

plt.tight_layout()
plt.show()
```

```python
# Age vs Diabetes

fig, axes = plt.subplots(1, 2, figsize=(12,5))

# Graph
sns.boxplot(x='Diabetes_012', y='Age', data=df, ax=axes[1])
axes[1].set_title("Age vs Diabetes")

# Table
summary = df.groupby('Diabetes_012')['Age'].agg(['mean', 'median', 'std'])
axes[0].axis('off')
axes[0].table(
    cellText=summary.round(2).values,
    colLabels=summary.columns,
    rowLabels=summary.index,
    loc='center'
)

axes[0].set_title("Age Summary Statistics")

plt.tight_layout()
plt.show()
```

```python
# Physical activity vs Diabetes

fig, axes = plt.subplots(1, 2, figsize=(14,5))

# Graph
sns.countplot(x='PhysActivity', hue='Diabetes_012', data=df, ax=axes[1])
axes[1].set_title("Physical Activity vs Diabetes")

# Table (percentages)
table = pd.crosstab(df['PhysActivity'], df['Diabetes_012'], normalize='index') * 100
axes[0].axis('off')
axes[0].table(
    cellText=table.round(2).values,
    colLabels=table.columns,
    rowLabels=table.index,
    loc='center'
)

axes[0].set_title("Proportions (%)")

plt.tight_layout()
plt.show()
```

```python
# Hypertension vs Diabetes
plt.figure()
sns.countplot(x='HighBP', hue='Diabetes_012', data=df)
plt.title("Hypertension vs Diabetes")
plt.show()
```

```python
# CORRELATION ANALYSIS
```

```python
plt.figure(figsize=(10,8))
corr = df.corr()

sns.heatmap(corr, annot=False)
plt.title("Correlation Matrix")
plt.show()
```

```python
# HIGH-RISK PROFIL
sns.scatterplot(x='Age', y='BMI', hue='Diabetes_012', data=df)
plt.title("High Risk Profiles (Age vs BMI)")
plt.show()
```
